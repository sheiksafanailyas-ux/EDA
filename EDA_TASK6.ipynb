```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"])
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"])
df.dtypes
```

```python
df_pivot=pd.pivot_table(
    df,
    index="Medical Condition",
    values="Billing Amount",
    columns="Insurance Provider",
    aggfunc="sum"
)
print(df_pivot)
```

```python
df_pivot.plot(kind="bar",stacked=True,figsize=(12,6))
plt.title("Comparing Billing Amount By Medical Condition And Insurance Provider")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=90)
plt.legend()
plt.show()
```

```python
sns.violinplot(x="Insurance Provider", y="Billing Amount", data=df)

plt.title("Billing Amount By Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Billing Amount")
plt.xticks(rotation=90)
plt.show()
```

```python
df["Hospital stay days"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days

df["Hospital stay days"]
```

```python
numeric_cols=["Age","Billing Amount","Hospital stay days"]
correlation=df[numeric_cols].corr()
correlation
```

```python

sns.heatmap(correlation,
            annot=True,
            cmap="Green",
            cbar=False,
            square=True,
            linewidth=1)
plt.title("correlation")
plt.show()
```
