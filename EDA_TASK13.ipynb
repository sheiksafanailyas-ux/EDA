```python
# Check the missing values
print(df[["country", "agent", "company"]].isnull().sum())

# Fill the missing values
df["country"] = df["country"].fillna(df["country"].mode()[0])
df["agent"] = df["agent"].fillna("Unknown")
df["company"] = df["company"].fillna("Unknown")

# Check again
print(df[["country", "agent", "company"]].isnull().sum())
```

```python
# Total length of stay
df["total_length"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Replace missing children values with 0
df["children"] = df["children"].fillna(0)

# Total number of guests
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

print(df[["total_length", "total_guests"]].head(10))
```

```python
cancellation=df.groupby("hotel")["is_canceled"].agg(
    ["sum","count"]
)
cancellation["cancillation_percentage"]=(cancellation["sum"]/cancellation["count"]
)*100
print(cancellation)
```

```python
def detect_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print("column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
    print()

    return outliers
```

```python
lead_time_outliers=detect_outliers("lead_time")
adr_outliers=detect_outliers("adr")
```
