```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"],errors='coerce')
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"],errors='coerce')
```

```python
def categorize_urgency(admission):
    if admission=='Emergency':
        return'Immediate care'
    elif admission=='Urgent':
        return'high priority'
    else:
        return'planned care'

df['Priority_level']=df['Admission Type'].apply(categorize_urgency)
```

```python
df
```

```python
#to find the hospital stay days we have to subtract Discharde date and Date of Admission
df["Hospital stay days"]=(df["Discharge Date"]-df["Date of Admission"]).dt.days
```
