```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("StudentsPerformance1.csv")
```

```python


parental_avg=df.groupby("parental level of education")[
    ["math score","reading score","writing score"]
    ].mean()
print("\n Average parental level of education")
print(parental_avg)

parental_avg.plot(
    kind="bar",
    figsize=(8,6)
)
plt.title("Average test score by parental level of education")
plt.xlabel("parental education level")
plt.ylabel("Average score")
plt.xticks(rotation=45)
plt.legend(title="subject")
plt.tight_layout()
plt.show()
```

```python
lunch_avg=df.groupby("lunch")[
    ["math score","reading score","writing score"]
    ].mean()
print("\n Average score by lunch")
print(lunch_avg)

lunch_avg.plot(
    kind="bar",
    figsize=(8,6)
)
plt.title("Average test score by lunch type")
plt.xlabel("lunch type")
plt.ylabel("Average score")
plt.xticks(rotation=45)
plt.legend(title="subject")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(
    data=df,
    x="test preparation course",
    y="reading score"
)

plt.title("Reading score by test preparation course")
plt.xlabel("test preparation course")
plt.ylabel("Reading score")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(
    data=df,
    x="test preparation course",
    y="math score"
)

plt.title("Reading score by test preparation course")
plt.xlabel("test preparation course")
plt.ylabel("math score")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(
    data=df,
    x="test preparation course",
    y="writing score"
)

plt.title("Reading score by test preparation course")
plt.xlabel("test preparation course")
plt.ylabel("writing  score")
plt.show()
```
