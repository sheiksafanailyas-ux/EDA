```python
df["Price_Delta"] = df["Close"] - df["Open"]
df
```

```python
df["Daily_return"] = df["Close"].pct_change() * 100
```

```python
df["Volume_MA_7"] = df["Volume"].rolling(7).mean()
```

```python
volume_mean = df["Volume"].mean()
volume_std = df["Volume"].std()
volume_threshold = volume_mean + (2 * volume_std)
```

```python
anomalies = df[df["Volume"] > volume_threshold]
print(anomalies)
```

```python
mean_return = df["Daily_return"].mean()
print("Mean Return:", mean_return)
```

```python
variance_return = df["Daily_return"].var()
print("Variance Return:", variance_return)
```

```python
std_return = df["Daily_return"].std()
print("Std Return:", std_return)
```
