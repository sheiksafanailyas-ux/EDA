```python
print(df.groupby('Country').mean(numeric_only=True).round(2))
```

```python
print(df.groupby('Water Scarcity Level').mean(numeric_only=True).round(2))
```

```python
print(df.groupby('Year').mean(numeric_only=True).round(2))
```

```python
df.groupby('Country')['Total Water Consumption (Billion Cubic Meters)'].mean().sort_values().plot(kind='barh')
plt.title('1. Average Total Consumption by Country')
plt.show()
```

```python
df.groupby('Year')['Total Water Consumption (Billion Cubic Meters)'].mean().plot(marker='o')
plt.title('2. Yearly Trend of Total Consumption')
plt.show()
```

```python
df.groupby('Country')[['Agricultural Water Use (%)', 'Industrial Water Use (%)', 'Household Water Use (%)']].mean().plot(kind='bar', stacked=True)
plt.title('3. Water Use by Sector')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(x='Water Scarcity Level', y='Per Capita Water Use (Liters per Day)', data=df)
plt.title('4. Per Capita by Scarcity')
plt.show()
```

```python
plt.figure(figsize=(12,8 ))
sns.heatmap(df.corr(numeric_only=True), 
            annot=True, 
            cmap='coolwarm',      
            cbar=True,          
            linewidths=0.5,       
)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
df.groupby('Country')['Groundwater Depletion Rate (%)'].mean().sort_values(ascending=False).plot(kind='bar')
plt.title('6. Groundwater Depletion by Country')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(df['Rainfall Impact (Annual Precipitation in mm)'], 
            df['Total Water Consumption (Billion Cubic Meters)'])
plt.title('7. Rainfall vs Total Consumption')
plt.xlabel('Rainfall')
plt.ylabel('Total Consumption')
plt.show()
```

```python
df.groupby('Year')['Per Capita Water Use (Liters per Day)'].mean().plot(marker='s')
plt.title('8. Per Capita Trend Over Years')
plt.show()
```
